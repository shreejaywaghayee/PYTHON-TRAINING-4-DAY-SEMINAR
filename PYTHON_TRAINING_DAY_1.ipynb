```python
#reverse the string without using any shortcut and in built function

s = "Learning python is very easy"
reversed_string = ""
current_word = ""
for char in s:
    if char == "":
        reversed_string = reversed_string + " " + current_word
        current_word = ""
    else:
        current_word += char
reversed_string = current_word + " " + reversed_string
print(reversed_string)
```

```python
class MyPython:
    def __init__(self):
        self.a = 10            # del function can be used to delete variable and constructor as well
        self.b = 20
        self.c = 30
        self.d = 40

    def func(self):
        del self.d  # Deletes instance variable 'd'

n = MyPython()
print(n.__dict__)   # Fixed to n.__dict__

n.func()
print(n.__dict__)   # 'd' will be gone here

del n.c             # Fixed to del n.c
print(n.__dict__)   # 'c' will be gone here
```

```python
#Accept 5 paper marks ,roll no ,name; and calculate total and percentage ifpercentage is >=85,student can take admission else cant take admission , using class and object

class StudentAdmission:
    def __init__(self):
        self.roll_no = ""
        self.name = ""
        self.marks = []
        self.total = 0
        self.percentage = 0.0
    def accept_details(self):
        self.roll_no = input("Enter Roll Number: ")
        self.name = input("Enter Name: ")
        print("Enter marks for 5 papers (out of 100 each):")
        self.marks = []
        for i in range(1, 6):
            m = float(input(f"Paper {i} marks: "))
            self.marks.append(m)
    def calculate_score(self):
        self.total = sum(self.marks)
        self.percentage = (self.total / 500) * 100
    def check_admission(self):
        print("- Admission Report 55 -")
        print("Roll No:", self.roll_no)
        print("Name:", self.name)
        print("Total Marks:", self.total)
        print(f"Percentage: {self.percentage:.2f}%")

        if self.percentage >= 85:
            print("Result: Eligible! Student can take admission.")
        else:
            print("Result: Not Eligible! Student cant take admission.")
student = StudentAdmission()
student.accept_details()
student.calculate_score()
student.check_admission()
```

```python
#Single Inheritance: A type of inheritance where a single child class inherits properties and methods from a single parent class.
#Multiple Inheritance: A type of inheritance where a single child class inherits from more than one parent class (supported in Python, but not in languages like Java).
#Hierarchical Inheritance: A type of inheritance where multiple child classes inherit from a single parent class.
#Hybrid Inheritance: A combination of two or more types of inheritance (for example, combining Multilevel and Multiple inheritance).
```

```python
class A:
    def showA(self):
        print("showA function")

class B(A):
    def showB(self):
        print("showB function")

class C(B):
    def showC(self):  # Defined as a method correctly
        print("showC function")

obj = C()
obj.showA()
obj.showB()
obj.showC()
```

```python
#multiple irrators
class A:
    def showA(self):
        print("showA function")

class B:
    def showB(self):
        print("showB function")

class C(A,B):
    def showC(self):  # Defined as a method correctly
        print("showC function")

obj = C()
obj.showA()
obj.showB()
obj.showC()
```

```python
class A:
    def showA(self):
        print("showA function")

class B(A):
    def showB(self):
        print("showB function")

class C(A):
    def showC(self):
        print("showC function")

print(" Object B ")
obj1 = B()
obj1.showA()
obj1.showB()
print("Object C")
obj2 = C()
obj2.showA()
obj2.showC()
```

```python
#hybrid inheritance

class A:
    def showA(self):
        print("showA function")

class B(A):
    def showB(self):
        print("showB function")

class C(A):
    def showC(self):
        print("showC function")

class D(B, C):  # Inherits from both B and C
    def showD(self):
        print("showD function")

obj = D()
obj.showA()
obj.showB()
obj.showC()
obj.showD()
```
