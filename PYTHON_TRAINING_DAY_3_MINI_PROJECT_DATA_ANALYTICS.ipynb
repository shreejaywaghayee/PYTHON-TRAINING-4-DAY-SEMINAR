```python
import pandas as pd
data = {
    'Name': ['A', 'B', 'C', 'D', 'E'],
    'Math': [78, 67, 90, 56, 88],
    'Science': [85, 72, 95, 60, 84],
    'English': [88, 79, 92, 58, 86]
}
df = pd.DataFrame(data)
print(df)
```

```python
import numpy as np

# Corrected 'Maths' to 'Math' to match the DataFrame column name
marks = df[['Math', 'Science', 'English']].values
average_marks = np.mean(marks, axis=1)
df['Average'] = average_marks
print(df)
```

```python
def grade(avg):
  if avg >=90:
    return 'A'
  elif avg >=75:
    return 'B'
  elif avg >=60:
    return 'C'
  else:
    return 'D'

  df['Grade']=df['Average'].apply(grade)
  df
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.bar(df["Name"], df["Average"])
plt.xlabel('Students')
plt.ylabel('Average Marks')
plt.title('Average Marks of students')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Average marks per subject
df[["Math", "Science", "English"]].mean().plot(kind="bar")
plt.title("Average Marks per Subject")
plt.ylabel("Marks")
plt.show()
```
